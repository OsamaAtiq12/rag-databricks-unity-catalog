```python
# Optional if sentence-transformers is not on the cluster:
# %pip install sentence-transformers

import re
import numpy as np
from pyspark.sql import Row

from sentence_transformers import SentenceTransformer

CATALOG = "25280091_pa3"


# ---------- 1) Fixed-size character splitter ----------
class MyTextSplitter:
    def __init__(self, chunk_size: int = 512, chunk_overlap: int = 50):
        if chunk_overlap >= chunk_size:
            raise ValueError("chunk_overlap must be smaller than chunk_size")
        self.chunk_size = chunk_size
        self.chunk_overlap = chunk_overlap

    def split(self, text: str) -> list[str]:
        if not text:
            return []
        chunks: list[str] = []
        start = 0
        n = len(text)
        while start < n:
            end = min(start + self.chunk_size, n)
            chunks.append(text[start:end])
            if end == n:
                break
            start = end - self.chunk_overlap
        return chunks


# ---------- 2) Semantic splitter (sentence + buffer + embeddings + cosine distance) ----------
def _cosine_sim(a: np.ndarray, b: np.ndarray) -> float:
    denom = (np.linalg.norm(a) * np.linalg.norm(b)) + 1e-12
    return float(np.dot(a, b) / denom)


class MySemanticSplitter:
    def __init__(self, distance_threshold: float = 0.15, buffer_size: int = 1):
        self.distance_threshold = distance_threshold
        self.buffer_size = buffer_size

    def _split_sentences(self, text: str) -> list[str]:
        parts = re.split(r"(?<=[.!?])\s+", (text or "").strip())
        return [p.strip() for p in parts if p.strip()]

    def _get_embeddings(self, texts: list[str]) -> np.ndarray:
        raise NotImplementedError

    def split(self, text: str) -> list[str]:
        sentences = self._split_sentences(text)
        if not sentences:
            return []
        if len(sentences) == 1:
            return [" ".join(sentences)]

        padded: list[str] = []
        for i in range(len(sentences)):
            lo = max(0, i - self.buffer_size)
            hi = min(len(sentences), i + self.buffer_size + 1)
            padded.append(" ".join(sentences[lo:hi]))

        emb = self._get_embeddings(padded)
        emb = np.asarray(emb, dtype=np.float64)
        if emb.shape[0] != len(sentences):
            raise ValueError("embedding count must match sentence count")

        chunks: list[str] = []
        chunk_start = 0
        for i in range(len(sentences) - 1):
            sim = _cosine_sim(emb[i], emb[i + 1])
            dist = 1.0 - sim
            if dist > self.distance_threshold:
                chunks.append(" ".join(sentences[chunk_start : i + 1]))
                chunk_start = i + 1
        chunks.append(" ".join(sentences[chunk_start:]))
        return [c for c in chunks if c.strip()]


class MySemanticSplitterST(MySemanticSplitter):
    def __init__(
        self,
        distance_threshold: float = 0.15,
        buffer_size: int = 1,
        model_name: str = "all-MiniLM-L6-v2",
    ):
        super().__init__(distance_threshold=distance_threshold, buffer_size=buffer_size)
        self._model = SentenceTransformer(model_name)

    def _get_embeddings(self, texts: list[str]) -> np.ndarray:
        return np.asarray(self._model.encode(texts, show_progress_bar=False))


# ---------- 3) Read knowledge base, chunk, write Delta tables ----------
kb = spark.table(f"`{CATALOG}`.`knowledge_base_data`.`knowledge_base`")

# Fixed: size=512, overlap=50
fixed_splitter = MyTextSplitter(chunk_size=512, chunk_overlap=50)
fixed_rows: list[Row] = []
for r in kb.collect():
    kb_id = r["Id"]
    content = r["Content"] or ""
    for idx, ch in enumerate(fixed_splitter.split(content)):
        fixed_rows.append(Row(knowledge_base_id=kb_id, chunk_index=idx, chunk_text=ch))

fixed_df = spark.createDataFrame(fixed_rows)
fixed_df.write.mode("overwrite").format("delta").saveAsTable(
    f"`{CATALOG}`.`processed_data`.`fixed_chunks`"
)

# Semantic: distance_threshold=0.15, buffer=1
sem_splitter = MySemanticSplitterST(distance_threshold=0.15, buffer_size=1)
sem_rows: list[Row] = []
for r in kb.collect():
    kb_id = r["Id"]
    content = r["Content"] or ""
    for idx, ch in enumerate(sem_splitter.split(content)):
        sem_rows.append(Row(knowledge_base_id=kb_id, chunk_index=idx, chunk_text=ch))

sem_df = spark.createDataFrame(sem_rows)
sem_df.write.mode("overwrite").format("delta").saveAsTable(
    f"`{CATALOG}`.`processed_data`.`semantic_chunks`"
)

print("fixed_chunks:", fixed_df.count(), "rows")
print("semantic_chunks:", sem_df.count(), "rows")
```

```python
# If needed:
# %pip install sentence-transformers
# dbutils.library.restartPython()

import numpy as np
from pyspark.sql import SparkSession
from pyspark.sql.types import StructType, StructField, StringType, ArrayType, FloatType
from sentence_transformers import SentenceTransformer

CATALOG = "25280091_pa3"


class GenerateEmbeddings:
    """
    Reads a chunk table, embeds a text column, writes Delta with:
      ChunkText STRING, Embedding ARRAY<FLOAT>
    """

    def __init__(self, model_name: str = "all-MiniLM-L6-v2"):
        self.model_name = model_name
        self._model = None

    def _get_model(self) -> SentenceTransformer:
        if self._model is None:
            self._model = SentenceTransformer(self.model_name)
        return self._model

    def run(
        self,
        spark: SparkSession,
        source_table: str,
        target_table: str,
        text_col: str = "chunk_text",
        batch_size: int = 64,
    ) -> None:
        model = self._get_model()

        df = spark.table(source_table)
        # Small/medium homework tables: collect texts (simple and fine for PA scale)
        texts = [row[text_col] for row in df.select(text_col).collect()]

        if len(texts) == 0:
            schema = StructType(
                [
                    StructField("ChunkText", StringType(), True),
                    StructField("Embedding", ArrayType(FloatType(), containsNull=False), True),
                ]
            )
            spark.createDataFrame([], schema).write.mode("overwrite").format("delta").saveAsTable(
                target_table
            )
            return

        vecs = model.encode(
            texts,
            batch_size=batch_size,
            show_progress_bar=True,
            normalize_embeddings=False,
        )
        vecs = np.asarray(vecs, dtype=np.float32)

        rows = [
            (texts[i], vecs[i].astype(float).tolist())  # Array[Float]
            for i in range(len(texts))
        ]

        schema = StructType(
            [
                StructField("ChunkText", StringType(), True),
                StructField("Embedding", ArrayType(FloatType(), containsNull=False), True),
            ]
        )

        out_df = spark.createDataFrame(rows, schema)
        out_df.write.mode("overwrite").format("delta").saveAsTable(target_table)


# ---------- Run on both chunk streams ----------
ge = GenerateEmbeddings(model_name="all-MiniLM-L6-v2")

ge.run(
    spark,
    source_table=f"`{CATALOG}`.`processed_data`.`fixed_chunks`",
    target_table=f"`{CATALOG}`.`processed_data`.`fixed_embeddings`",
    text_col="chunk_text",
)

ge.run(
    spark,
    source_table=f"`{CATALOG}`.`processed_data`.`semantic_chunks`",
    target_table=f"`{CATALOG}`.`processed_data`.`semantic_embeddings`",
    text_col="chunk_text",
)

print("fixed_embeddings rows:", spark.table(f"`{CATALOG}`.`processed_data`.`fixed_embeddings`").count())
print("semantic_embeddings rows:", spark.table(f"`{CATALOG}`.`processed_data`.`semantic_embeddings`").count())
```

```python
import logging
import warnings
import numpy as np
from sentence_transformers import SentenceTransformer
logging.getLogger("transformers").setLevel(logging.ERROR)
logging.getLogger("sentence_transformers").setLevel(logging.ERROR)
warnings.filterwarnings("ignore", module="transformers")
def _cosine_similarity(q: np.ndarray, v: np.ndarray) -> float:
    denom = float(np.linalg.norm(q) * np.linalg.norm(v)) + 1e-12
    return float(np.dot(q, v) / denom)
class VectorSearch:
    def __init__(
        self,
        embeddings_table: str,
        model_name: str = "all-MiniLM-L6-v2",
        text_col: str = "ChunkText",
        embedding_col: str = "Embedding",
        model: SentenceTransformer | None = None,
    ):
        self.embeddings_table = embeddings_table
        self.text_col = text_col
        self.embedding_col = embedding_col
        self._model_name = model_name
        self._model = model
    def _get_model(self) -> SentenceTransformer:
        if self._model is None:
            self._model = SentenceTransformer(self._model_name)
        return self._model
    def search(self, query: str, top_k: int = 5):
        if not (query or "").strip():
            return []
        model = self._get_model()
        q = np.asarray(model.encode([query], show_progress_bar=False)[0], dtype=np.float64)
        rows = spark.table(self.embeddings_table).select(self.text_col, self.embedding_col).collect()
        scored = []
        for r in rows:
            t = r[self.text_col]
            emb = r[self.embedding_col]
            if emb is None:
                continue
            vec = np.asarray(emb, dtype=np.float64)
            scored.append((_cosine_similarity(q, vec), t or ""))
        scored.sort(key=lambda x: x[0], reverse=True)
        return [{"Similarity": float(s), "Chunk_text": txt} for s, txt in scored[:top_k]]
_shared_embedder = SentenceTransformer("all-MiniLM-L6-v2")
vs_fixed = VectorSearch(
    f"`{CATALOG}`.`processed_data`.`fixed_embeddings`",
    model=_shared_embedder,
)
vs_sem = VectorSearch(
    f"`{CATALOG}`.`processed_data`.`semantic_embeddings`",
    model=_shared_embedder,
)
```

```python
from langchain_openai import AzureChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
class RAGPipeline:
    def __init__(self, vector_search: VectorSearch, llm=None, top_k: int = 5):
        self.vector_search = vector_search
        self.top_k = top_k
        self.llm = llm or AzureChatOpenAI(
            azure_deployment=AZURE_OPENAI_CHAT_DEPLOYMENT,
            api_version=AZURE_OPENAI_API_VERSION,
            azure_endpoint=AZURE_OPENAI_ENDPOINT,
            api_key=AZURE_OPENAI_API_KEY,
            temperature=0,
        )
        self.prompt = ChatPromptTemplate.from_messages(
            [
                (
                    "system",
                    "You are a careful assistant. Answer ONLY using the context below.\n"
                    "If the context does not contain enough information to answer the question, "
                    'reply exactly: I do not know\n'
                    "Do not use outside knowledge or guess.\n\n"
                    "Context:\n{context}",
                ),
                ("human", "{question}"),
            ]
        )
        self._chain = self.prompt | self.llm | StrOutputParser()
    def Get_context(self, question: str) -> str:
        hits = self.vector_search.search(question, top_k=self.top_k)
        return "\n\n".join(h["Chunk_text"] for h in hits if h.get("Chunk_text"))
    def Invoke(self, question: str) -> str:
        ctx = self.Get_context(question)
        msgs = self.prompt.format_messages(context=ctx, question=question)
        return (self.llm.invoke(msgs).content or "").strip()
    def Invoke_w_lcel(self, question: str) -> str:
        ctx = self.Get_context(question)
        return self._chain.invoke({"context": ctx, "question": question}).strip()
```

### Task 5 — Fixed vs semantic chunking (RAG answers)

**Question:** *What are search agents and how do they differ from traditional RAG pipelines?*

**Observed outputs:**  
- **Fixed chunks (`Invoke`):** Describes search agents (RAG 2.0) as interleaving reasoning and retrieval, contrasts with sequential retrieve-then-generate RAG, and notes decisions about *when/what* to retrieve follow ongoing reasoning.  
- **Semantic chunks (`Invoke_w_lcel`):** Delivers the same technical contrast with nearly the same structure; phrasing shifts slightly (e.g., “advanced paradigm” vs explicitly naming “RAG 2.0” in the opening).

**Do the answers differ?**  
**Only mildly.** There is no disagreement in facts; differences are mainly **wording** and **emphasis**.

**Why they are alike**  
Same **LLM**, **prompt constraints**, and **document**; retrieved context overlaps on the paper’s core definition (dynamic interleaving vs fixed pipelines), so the synthesized answers converge.

**Why they can differ**  
**`Get_context`** pulls different chunk boundaries (fixed-size windows vs semantic splits), so the model may **surface slightly different supporting sentences** (e.g., whether “knowledge bases” is emphasized in the last sentence).

**Invocation note:** `Invoke` and `Invoke_w_lcel` are alternative LangChain execution styles; with **temperature 0**, they should not be the main source of divergence—**retrieval** is.

```python
import time
import pprint

from databricks.vector_search.client import VectorSearchClient

# ===================== CONFIG =====================
CATALOG = "25280091_pa3"
ROLL = "25280091"

SOURCE_TABLE = f"{CATALOG}.processed_data.fixed_chunks_prepared"
INDEX_NAME = f"{CATALOG}.vector_index.fixed_vector_index"
EMBEDDING_MODEL = "databricks-gte-large-en"

# If you know the only/shared endpoint name, set it here to skip parsing.
ENDPOINT_NAME_OVERRIDE = None  # e.g. "my_shared_vector_endpoint"

# ===================== CLIENT =====================
vsc = VectorSearchClient(disable_notice=True)


def _unwrap_endpoints(raw):
    """Normalize list_endpoints() return value to a list of endpoint records."""
    if raw is None:
        return []

    # Common: {"endpoints": [ {...}, ... ]}
    if isinstance(raw, dict):
        if isinstance(raw.get("endpoints"), list):
            return raw["endpoints"]
        if isinstance(raw.get("data"), list):
            return raw["data"]
        if "name" in raw:
            return [raw]
        # Fallback: first list-valued key
        for v in raw.values():
            if isinstance(v, list):
                return v
        return []

    if isinstance(raw, list):
        return raw

    return [raw]


def _endpoint_name_from_record(ep):
    if ep is None:
        return None
    if isinstance(ep, str):
        return ep
    if isinstance(ep, dict):
        return ep.get("name") or ep.get("endpoint_name")
    return getattr(ep, "name", None) or getattr(ep, "endpoint_name", None)


def pick_or_resolve_endpoint_name():
    if ENDPOINT_NAME_OVERRIDE:
        return ENDPOINT_NAME_OVERRIDE

    raw = vsc.list_endpoints()
    items = _unwrap_endpoints(raw)
    names = [_endpoint_name_from_record(x) for x in items]
    names = [n for n in names if n]

    if not names:
        print("DEBUG list_endpoints() raw:")
        pprint.pprint(raw)
        raise RuntimeError("No endpoint names parsed. Check DEBUG output above.")

    if len(names) > 1:
        print("Multiple endpoints; using [0]. Set ENDPOINT_NAME_OVERRIDE to pick one:")
        pprint.pprint(names)

    return names[0]


def _get_endpoint(name: str):
    try:
        return vsc.get_endpoint(name=name)
    except TypeError:
        return vsc.get_endpoint(name)


def _endpoint_status_upper(ep) -> str:
    if isinstance(ep, dict):
        d = ep
    elif hasattr(ep, "__dict__"):
        d = ep.__dict__
    else:
        return str(ep).upper()

    for k in ("endpoint_status", "state", "status"):
        if k in d:
            return str(d[k]).upper()
    return str(d).upper()


def wait_for_endpoint_ready(name: str, max_wait_s: int = 1800, sleep_s: int = 15):
    deadline = time.time() + max_wait_s
    while time.time() < deadline:
        ep = _get_endpoint(name)
        s = _endpoint_status_upper(ep)
        print(time.strftime("%H:%M:%S"), "endpoint status:", s[:400])
        if any(x in s for x in ("ONLINE", "READY", "PROVISIONING_COMPLETED", "RUNNING")):
            return ep
        time.sleep(sleep_s)
    pprint.pprint(_get_endpoint(name))
    raise TimeoutError(f"Endpoint {name} not ready in {max_wait_s}s")


def _get_index(endpoint_name: str, index_name: str):
    try:
        return vsc.get_index(endpoint_name=endpoint_name, index_name=index_name)
    except TypeError:
        return vsc.get_index(endpoint_name, index_name)


def wait_for_index_online(endpoint_name: str, index_name: str, max_wait_s: int = 7200, sleep_s: int = 30):
    deadline = time.time() + max_wait_s
    idx = _get_index(endpoint_name, index_name)
    while time.time() < deadline:
        desc = idx.describe()
        status = desc.get("status", {}) if isinstance(desc, dict) else {}
        detailed = str(status.get("detailed_state", "")).upper()
        ready = status.get("ready")
        msg = status.get("message", "")
        print(time.strftime("%H:%M:%S"), "index ready=", ready, "state=", detailed, "msg=", str(msg)[:200])
        if ready is True or detailed.startswith("ONLINE"):
            return desc
        time.sleep(sleep_s)
        idx = _get_index(endpoint_name, index_name)
    raise TimeoutError(f"Index {index_name} not online in {max_wait_s}s")


# ===================== RUN =====================
ENDPOINT_NAME = pick_or_resolve_endpoint_name()
print("Using endpoint:", ENDPOINT_NAME)

wait_for_endpoint_ready(ENDPOINT_NAME)

try:
    index = vsc.create_delta_sync_index(
        endpoint_name=ENDPOINT_NAME,
        source_table_name=SOURCE_TABLE,
        index_name=INDEX_NAME,
        pipeline_type="TRIGGERED",
        primary_key="Chunk_Id",
        embedding_source_column="chunk_text",
        embedding_model_endpoint_name=EMBEDDING_MODEL,
    )
    print("Index create requested:", INDEX_NAME)
    try:
        display(index.describe())
    except Exception:
        pprint.pprint(index.describe())
except TypeError as e:
    import inspect
    print("Fix kwargs to match your SDK:", e)
    print(inspect.signature(vsc.create_delta_sync_index))
    raise
except Exception as e:
    msg = str(e).lower()
    if "already exists" in msg or "resource_already_exists" in msg:
        print("Index likely already exists — describing:")
        try:
            display(_get_index(ENDPOINT_NAME, INDEX_NAME).describe())
        except Exception:
            pprint.pprint(_get_index(ENDPOINT_NAME, INDEX_NAME).describe())
    else:
        raise

wait_for_index_online(ENDPOINT_NAME, INDEX_NAME)
print("DONE: index ready/online.")
```

```python
rag_prompt = ChatPromptTemplate.from_messages([
    ("system",
     "You are a careful assistant. Answer ONLY using the context below.\n"
     "If the context is insufficient, reply exactly: I do not know\n"
     "Context:\n{context}"),
    ("human", "{question}"),
])
grader_prompt = ChatPromptTemplate.from_messages([
    ("system",
     "You are a strict grader. Decide if the provided context is sufficient.\n"
     "Respond with exactly one token: YES or NO.\n"
     "Context:\n{context}"),
    ("human", "{question}"),
])
fallback_prompt = ChatPromptTemplate.from_messages([
    ("system",
     "You are a careful assistant. Use ONLY the fallback document below. "
     "If it lacks the answer, reply exactly: I do not know\n"
     "Fallback document:\n{fallback_text}"),
    ("human", "{question}"),
])

from langchain_core.output_parsers import StrOutputParser
rag_chain = rag_prompt | llm | StrOutputParser()
grader_chain = grader_prompt | llm | StrOutputParser()
fallback_chain = fallback_prompt | llm | StrOutputParser()
```

```python
def retrieve_formatted_context(question: str, top_k: int = TOP_K):
    res = vs_index.similarity_search(
        query_text=question,
        columns=["chunk_text"],
        num_results=top_k,  # SDK expects num_results
    )
    hits = res.get("result", {}).get("data_array", []) if isinstance(res, dict) else []
    return [h[0] for h in hits if h and h[0]]
```

```python
from typing import TypedDict, List, Optional, Literal

class GraphState(TypedDict):
    question: str
    context: str
    answer: str
    grade: Optional[bool]
    route: Optional[Literal["vector", "fallback"]]
    retrieved_chunks: List[str]

def context_retriever_node(state: GraphState) -> GraphState:
    chunks = retrieve_formatted_context(question=state["question"], top_k=TOP_K)
    return {**state, "retrieved_chunks": chunks, "context": "\n\n".join(chunks)}

def grader_node(state: GraphState) -> GraphState:
    decision = grader_chain.invoke({"context": state.get("context", ""), "question": state["question"]})
    ok = str(decision).strip().upper().startswith("Y")
    return {**state, "grade": ok}

def base_rag_node(state: GraphState) -> GraphState:
    ans = rag_chain.invoke({"context": state.get("context", ""), "question": state["question"]}).strip()
    return {**state, "answer": ans, "route": "vector"}

def text_file_fallback_node(state: GraphState) -> GraphState:
    fb_df = spark.read.text(FALLBACK_TEXT_DIR)
    fb_text = "\n".join(r["value"] for r in fb_df.collect())
    ans = fallback_chain.invoke({"fallback_text": fb_text, "question": state["question"]}).strip()
    return {**state, "answer": ans, "route": "fallback", "context": fb_text}

def route_after_grade(state: GraphState) -> str:
    return "use_vector" if state.get("grade") else "use_fallback"
```

```python
from langgraph.graph import StateGraph, END

graph = StateGraph(GraphState)
graph.add_node("retrieve", context_retriever_node)
graph.add_node("grade", grader_node)
graph.add_node("rag", base_rag_node)
graph.add_node("fallback", text_file_fallback_node)

graph.set_entry_point("retrieve")
graph.add_edge("retrieve", "grade")
graph.add_conditional_edges("grade", route_after_grade, {
    "use_vector": "rag",
    "use_fallback": "fallback",
})
graph.add_edge("rag", END)
graph.add_edge("fallback", END)

agentic_rag_flow = graph.compile()
```

```python
def run_agentic_rag(question: str):
    init: GraphState = {
        "question": question,
        "context": "",
        "answer": "",
        "grade": None,
        "route": None,
        "retrieved_chunks": [],
    }
    result = agentic_rag_flow.invoke(init)
    print("Route:", result["route"])
    print("Grade OK?:", result["grade"])
    print("Retrieved (top 3):", result["retrieved_chunks"][:3])
    print("\nAnswer:\n", result["answer"])
    return result

# Example:
# run_agentic_rag("What are search agents and how do they use retrieval?")
```

```python
# Bonus tasks Start from here

import os, json
from typing import Optional
from langchain.tools import tool
from langgraph.checkpoint.memory import MemorySaver
from langgraph.prebuilt import create_react_agent
from langchain_openai import AzureChatOpenAI
from databricks.vector_search.client import VectorSearchClient

# Config (adjust if your names differ)
CATALOG = "25280091_pa3"
VECTOR_ENDPOINT = "25280091_vector_endpoint"
INDEX_NAME = f"{CATALOG}.vector_index.fixed_vector_index"
FALLBACK_DIR = f"/dbfs/Volumes/{CATALOG}/default/text_documents"  # expects databricks.txt / azure.txt inside

vsc = VectorSearchClient(disable_notice=True)
idx = vsc.get_index(endpoint_name=VECTOR_ENDPOINT, index_name=INDEX_NAME)

def retrieve_formatted_context(question: str, topk: int = 3) -> str:
    """Query vector index and return top-k chunk_text joined with separators."""
    res = idx.similarity_search(query_text=question, columns=["chunk_text"], num_results=topk)
    hits = res.get("result", {}).get("data_array", []) if isinstance(res, dict) else []
    raw_docs = [h[0] for h in hits if h and len(h) > 0 and h[0]]
    return "\n\n---\n\n".join(raw_docs)

@tool("vector_search_tool")
def vector_search_tool(query: str, top_k: int = 3) -> str:
    """Retrieve top-k chunks for a query from the fixed vector index."""
    ctx = retrieve_formatted_context(query, top_k)
    return ctx or "No results found"

@tool("read_fallback_document")
def read_fallback_document(category: str) -> str:
    """Read a fallback .txt document from the text_documents volume by category (e.g., databricks, azure)."""
    category = (category or "").strip().lower()
    file_map = {
        "databricks": os.path.join(FALLBACK_DIR, "databricks.txt"),
        "azure": os.path.join(FALLBACK_DIR, "azure.txt"),
    }
    file_path = file_map.get(category)
    if not file_path:
        return f"No documentation found for category: {category}"
    try:
        with open(file_path, "r", encoding="utf-8") as f:
            return f.read()
    except Exception as e:
        return f"Error reading document: {e}"

def build_stateful_mcp_agent():
    llm = AzureChatOpenAI(
        azure_endpoint=os.environ.get("AZURE_OPENAI_ENDPOINT"),
        azure_deployment=os.environ.get("AZURE_OPENAI_CHAT_DEPLOYMENT"),
        api_version=os.environ.get("AZURE_OPENAI_API_VERSION") or os.environ.get("OPENAI_API_VERSION"),
        api_key=os.environ.get("AZURE_OPENAI_API_KEY"),
        temperature=0,
    )
    tools = [vector_search_tool, read_fallback_document]
    memory = MemorySaver()
    return create_react_agent(llm, tools=tools, checkpointer=memory)

agent = build_stateful_mcp_agent()
```

```python
# Run a validation set through the stateful agent (robust to AIMessage responses)
def execute_mcp_queries(mcp_agent, queries: list, thread_id: str):
    config = {"configurable": {"thread_id": thread_id}}
    for query in queries:
        print(f"\nUser: {query}")
        payload = {"messages": [{"role": "user", "content": query}]}
        resp = mcp_agent.invoke(payload, config=config)

        # Normalize to a list of message-like objects
        if isinstance(resp, dict):
            msgs = resp.get("messages", [])
        elif hasattr(resp, "messages"):
            msgs = resp.messages
        elif isinstance(resp, (list, tuple)):
            msgs = resp
        else:
            msgs = [resp]

        final_message = ""
        if msgs:
            last = msgs[-1]
            content = getattr(last, "content", last)
            if isinstance(content, list):  # OpenAI-style parts
                final_message = "".join(
                    x.get("text", "") if isinstance(x, dict) else str(x) for x in content
                )
            else:
                final_message = str(content)
        print(f"Agent: {final_message}")

validation_queries = [
    "What tools do you have?",
    "What is searchAgent-X?",
    "What does this agent do specifically?",
    "Use your tool to search for information about Azure.",
    "Based on those search results, summarize its main features.",
    "What is Databricks?",
    "What core features differentiate both products?",
]

print("--- TESTING STATEFUL MCP AGENT ---")
execute_mcp_queries(agent, validation_queries, "mcp_test_session_01")
```
